```python
# =========================================================
# Cell 1: Locate the repository and load the full dataset
# =========================================================
#
# This cell:
# - works locally and in Google Colab;
# - finds the repository;
# - locates the raw Google subset CSV;
# - loads all rows without sampling.
# =========================================================

import sys
from pathlib import Path
import numpy as np
import pandas as pd


# Detect the execution environment.
IN_COLAB = "google.colab" in sys.modules


# Find the repository root.
if IN_COLAB:
    from google.colab import drive  # pyright: ignore[reportMissingImports]

    drive.mount("/content/drive", force_remount=False)

    REPO_ROOT = Path(
        "/content/drive/Shareddrives/CMPE 295A_B/subset"
    )
else:
    REPO_ROOT = Path.cwd()

    while (
        REPO_ROOT.name != "masters-project"
        and REPO_ROOT.parent != REPO_ROOT
    ):
        REPO_ROOT = REPO_ROOT.parent

    if REPO_ROOT.name != "masters-project":
        raise RuntimeError(
            "Could not find repository root named 'masters-project'."
        )


# Locate the raw dataset.
RAW_CSV_PATH = REPO_ROOT / "subset" / "borg_traces_data.csv"

if not RAW_CSV_PATH.exists():
    RAW_CSV_PATH = (
        REPO_ROOT
        / "data"
        / "raw"
        / "subset"
        / "borg_traces_data.csv"
    )

if not RAW_CSV_PATH.exists():
    raise FileNotFoundError(
        f"Raw dataset was not found at: {RAW_CSV_PATH}"
    )


# Load the complete dataset.
# Do not use nrows because the full dataset is required.
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 160)

df = pd.read_csv(
    RAW_CSV_PATH,
    low_memory=False
)


# Display basic information.
print("Repository root:", REPO_ROOT)
print("Raw dataset path:", RAW_CSV_PATH)
print("Running in Colab:", IN_COLAB)
print(
    f"Dataset size: "
    f"{df.shape[0]:,} rows × {df.shape[1]} columns"
)

print("\nColumns:")
print(df.columns.tolist())

display(df.head(3))
```

```python
# =========================================================
# Cell 2: Parse nested resource fields
# =========================================================
#
# This cell:
# - parses dictionary-like resource fields;
# - checks missing CPU and memory values;
# - extracts request, average, and peak resources;
# - preserves missing values;
# - does not filter rows or convert units.
# =========================================================

import ast


NESTED_COLUMNS = [
    "resource_request",
    "average_usage",
    "maximum_usage",
    "random_sample_usage",
]


# Convert a dictionary-like string into a Python dictionary.
def parse_nested(value):
    if value is None:
        return None

    if isinstance(value, float) and pd.isna(value):
        return None

    if isinstance(value, dict):
        return value

    if isinstance(value, str):
        try:
            parsed = ast.literal_eval(value)

            if isinstance(parsed, dict):
                return parsed

        except (ValueError, SyntaxError):
            pass

    return None


# Get one value from a parsed resource dictionary.
def get_resource_value(value, key):
    if isinstance(value, dict):
        return value.get(key)

    return None


# Parse and inspect each nested resource column.
parsed_nested = {}
nested_summary = []

for column in NESTED_COLUMNS:
    parsed = df[column].map(parse_nested)
    parsed_nested[column] = parsed

    cpu_values = parsed.map(
        lambda value: get_resource_value(value, "cpus")
    )

    memory_values = parsed.map(
        lambda value: get_resource_value(value, "memory")
    )

    nested_summary.append(
        {
            "source_column": column,
            "missing_or_invalid_dict": int(parsed.isna().sum()),
            "missing_cpu_value": int(cpu_values.isna().sum()),
            "missing_memory_value": int(memory_values.isna().sum()),
        }
    )


print("Nested resource-field summary:")
display(pd.DataFrame(nested_summary))


# Extract raw resource values.
df["raw_req_cpu"] = pd.to_numeric(
    parsed_nested["resource_request"].map(
        lambda value: get_resource_value(value, "cpus")
    ),
    errors="coerce",
)

df["raw_req_memory"] = pd.to_numeric(
    parsed_nested["resource_request"].map(
        lambda value: get_resource_value(value, "memory")
    ),
    errors="coerce",
)

df["raw_avg_cpu"] = pd.to_numeric(
    parsed_nested["average_usage"].map(
        lambda value: get_resource_value(value, "cpus")
    ),
    errors="coerce",
)

df["raw_avg_memory"] = pd.to_numeric(
    parsed_nested["average_usage"].map(
        lambda value: get_resource_value(value, "memory")
    ),
    errors="coerce",
)

df["raw_peak_cpu"] = pd.to_numeric(
    parsed_nested["maximum_usage"].map(
        lambda value: get_resource_value(value, "cpus")
    ),
    errors="coerce",
)

df["raw_peak_memory"] = pd.to_numeric(
    parsed_nested["maximum_usage"].map(
        lambda value: get_resource_value(value, "memory")
    ),
    errors="coerce",
)


RAW_RESOURCE_COLUMNS = [
    "raw_req_cpu",
    "raw_req_memory",
    "raw_avg_cpu",
    "raw_avg_memory",
    "raw_peak_cpu",
    "raw_peak_memory",
]


print("\nExtracted raw resource values:")
display(df[RAW_RESOURCE_COLUMNS].head())

print("\nMissing percentages after extraction:")
display(
    df[RAW_RESOURCE_COLUMNS]
      .isna()
      .mean()
      .mul(100)
      .sort_values(ascending=False)
)
```

```python
# =========================================================
# Cell 3: Review resource values and raw timestamps
# =========================================================
#
# This cell:
# - summarizes CPU and memory values;
# - checks negative resources;
# - checks peak versus average usage;
# - compares peak usage with requests;
# - reviews timestamp ranges;
# - compares possible timestamp units;
# - checks raw usage-window differences.
#
# No rows are removed and no units are converted.
# =========================================================


# ---------------------------------------------------------
# Resource value distributions
# ---------------------------------------------------------

resource_summary = (
    df[RAW_RESOURCE_COLUMNS]
      .describe(
          percentiles=[
              0.01,
              0.05,
              0.25,
              0.50,
              0.75,
              0.95,
              0.99,
              0.999,
          ]
      )
      .T
)

print("Resource value summary:")
display(resource_summary)


# ---------------------------------------------------------
# Negative resource-value checks
# ---------------------------------------------------------

negative_counts = (
    df[RAW_RESOURCE_COLUMNS]
      .lt(0)
      .sum()
      .sort_values(ascending=False)
)

print("\nNegative resource-value counts:")
display(negative_counts)


# ---------------------------------------------------------
# Average-versus-peak checks
# ---------------------------------------------------------

cpu_values = df[
    ["raw_avg_cpu", "raw_peak_cpu"]
].dropna()

memory_values = df[
    ["raw_avg_memory", "raw_peak_memory"]
].dropna()

peak_cpu_below_average = (
    cpu_values["raw_peak_cpu"]
    < cpu_values["raw_avg_cpu"]
).sum()

peak_memory_below_average = (
    memory_values["raw_peak_memory"]
    < memory_values["raw_avg_memory"]
).sum()

print(
    "\nPeak CPU below average CPU:",
    int(peak_cpu_below_average)
)

print(
    "Peak memory below average memory:",
    int(peak_memory_below_average)
)


# ---------------------------------------------------------
# Request-versus-peak comparison
# ---------------------------------------------------------
#
# These are diagnostic counts only.
# They are not final violation metrics because execution
# reconstruction and canonical evaluation are not complete.

cpu_request_values = df[
    ["raw_req_cpu", "raw_peak_cpu"]
].dropna()

memory_request_values = df[
    ["raw_req_memory", "raw_peak_memory"]
].dropna()

peak_cpu_above_request = (
    cpu_request_values["raw_peak_cpu"]
    > cpu_request_values["raw_req_cpu"]
).sum()

peak_memory_above_request = (
    memory_request_values["raw_peak_memory"]
    > memory_request_values["raw_req_memory"]
).sum()

print(
    "\nPeak CPU above requested CPU:",
    int(peak_cpu_above_request)
)

print(
    "Peak memory above requested memory:",
    int(peak_memory_above_request)
)


# ---------------------------------------------------------
# Raw timestamp ranges
# ---------------------------------------------------------

SECONDS_PER_DAY = 24 * 60 * 60

start_min = df["start_time"].min()
start_max = df["start_time"].max()
end_min = df["end_time"].min()
end_max = df["end_time"].max()

print("\nRaw timestamp ranges:")
print(f"start_time: {start_min:,} --> {start_max:,}")
print(f"end_time  : {end_min:,} --> {end_max:,}")


# ---------------------------------------------------------
# Compare possible timestamp units
# ---------------------------------------------------------

print("\nCandidate interpretations of the start-time span:")

timestamp_unit_divisors = {
    "nanoseconds": 1e9,
    "microseconds": 1e6,
    "milliseconds": 1e3,
    "seconds": 1.0,
}

for unit_name, divisor in timestamp_unit_divisors.items():
    span_days = (
        (start_max - start_min)
        / divisor
        / SECONDS_PER_DAY
    )

    print(
        f"{unit_name:>12}: "
        f"{span_days:,.4f} days"
    )


# ---------------------------------------------------------
# Raw usage-window differences
# ---------------------------------------------------------
#
# This is only the difference between raw end_time and
# raw start_time. It is not final execution runtime.

raw_usage_window_delta = (
    pd.to_numeric(df["end_time"], errors="coerce")
    - pd.to_numeric(df["start_time"], errors="coerce")
)

print("\nMost common raw usage-window differences:")
display(
    raw_usage_window_delta
      .value_counts()
      .head(20)
      .rename_axis("raw_difference")
      .reset_index(name="row_count")
)

print("\nRaw usage-window difference summary:")
display(
    raw_usage_window_delta.describe(
        percentiles=[
            0.01,
            0.05,
            0.50,
            0.95,
            0.99,
        ]
    )
)


# ---------------------------------------------------------
# Convert the raw difference under each candidate unit
# ---------------------------------------------------------

print("\nUsage-window duration under candidate units:")

for unit_name, divisor in timestamp_unit_divisors.items():
    print(
        f"{unit_name:>12}: "
        f"minimum = {raw_usage_window_delta.min() / divisor:g}, "
        f"median = {raw_usage_window_delta.median() / divisor:g}, "
        f"maximum = {raw_usage_window_delta.max() / divisor:g}"
    )


print(
    "\nNegative usage-window differences:",
    int((raw_usage_window_delta < 0).sum())
)

print(
    "Zero usage-window differences:",
    int((raw_usage_window_delta == 0).sum())
)
```

The `time` column represents an event timestamp and is separate from the
resource usage window defined by `start_time` and `end_time`.

Three rows contain the maximum 64-bit integer and are treated as invalid
event timestamps. They will be converted to missing values without removing
the rows.

Thirteen event timestamps occur after the latest usage end time. These rows
are retained because lifecycle event time does not need to fall inside the
resource measurement window.

```python
# =========================================================
# Cell 4: Check event-time anomalies
# =========================================================
#
# This cell checks whether the raw event-time column contains
# sentinel values or timestamps outside the observed trace range.
#
# It does not remove or modify any rows.
# =========================================================

INT64_MAX = 2**63 - 1

trace_end = pd.to_numeric(
    df["end_time"],
    errors="coerce"
).max()

event_time = pd.to_numeric(
    df["time"],
    errors="coerce"
)

is_int64_sentinel = event_time == INT64_MAX
is_after_trace_end = event_time > trace_end

print("Rows with Int64 maximum sentinel:")
print(int(is_int64_sentinel.sum()))

print("\nRows with event time after the latest usage end time:")
print(int(is_after_trace_end.sum()))

print("\nMissing event-time values:")
print(int(event_time.isna().sum()))

print("\nEvent-time summary excluding sentinel values:")
display(
    event_time[
        ~is_int64_sentinel
    ].describe(
        percentiles=[
            0.01,
            0.50,
            0.99,
        ]
    )
)

print("\nSample anomalous event-time rows:")
display(
    df.loc[
        is_int64_sentinel | is_after_trace_end,
        [
            "time",
            "start_time",
            "end_time",
            "event",
            "failed",
            "collection_id",
            "instance_index",
            "machine_id",
        ],
    ].head(20)
)
```

```python
# =========================================================
# Cell 5: Review event and failure values
# =========================================================
#
# This cell checks which lifecycle events are present and
# how they relate to the failed flag.
#
# =========================================================

print("Event counts:")
display(
    df["event"]
      .value_counts(dropna=False)
      .rename_axis("event")
      .reset_index(name="row_count")
)

print("\nFailed-value counts:")
display(
    df["failed"]
      .value_counts(dropna=False)
      .rename_axis("failed")
      .reset_index(name="row_count")
)

print("\nEvent versus failed cross-tabulation:")
display(
    pd.crosstab(
        df["event"],
        df["failed"],
        dropna=False
    )
)

print("\nMissing event values:")
print(int(df["event"].isna().sum()))

print("\nMissing failed values:")
print(int(df["failed"].isna().sum()))
```

```python
# =========================================================
# Cell 6: Check resource completeness by event
# =========================================================
#
# This cell checks which event types have missing request,
# average-usage, or peak-usage values.
#
# It does not remove or filter any event types.
# =========================================================

event_resource_summary = (
    df.groupby("event", dropna=False)
      .agg(
          row_count=("event", "size"),

          missing_req_cpu=(
              "raw_req_cpu",
              lambda values: values.isna().sum()
          ),

          missing_req_memory=(
              "raw_req_memory",
              lambda values: values.isna().sum()
          ),

          missing_avg_cpu=(
              "raw_avg_cpu",
              lambda values: values.isna().sum()
          ),

          missing_avg_memory=(
              "raw_avg_memory",
              lambda values: values.isna().sum()
          ),

          missing_peak_cpu=(
              "raw_peak_cpu",
              lambda values: values.isna().sum()
          ),

          missing_peak_memory=(
              "raw_peak_memory",
              lambda values: values.isna().sum()
          ),
      )
      .sort_values("row_count", ascending=False)
)

print("Resource completeness by event:")
display(event_resource_summary)
```

```python
# =========================================================
# Cell 7: Review observation and candidate execution keys
# =========================================================
#
# This cell checks:
# - whether raw observation rows are unique;
# - how many candidate groups exist;
# - whether candidate groups contain multiple machines
#   or multiple usage windows.
#
# This cell does not create final executions.
# =========================================================

candidate_key = [
    "collection_id",
    "instance_index",
]

observation_key = [
    "collection_id",
    "instance_index",
    "machine_id",
    "start_time",
    "end_time",
]


# Check uniqueness of the raw observation key.
observation_key_counts = (
    df.groupby(observation_key, dropna=False)
      .size()
)

print("Total raw rows:", len(df))
print("Unique observation keys:", len(observation_key_counts))
print(
    "Duplicated observation keys:",
    int((observation_key_counts > 1).sum())
)
print(
    "Maximum rows per observation key:",
    int(observation_key_counts.max())
)


# Summarize candidate collection-instance groups.
candidate_summary = (
    df.groupby(candidate_key, dropna=False)
      .agg(
          raw_row_count=("machine_id", "size"),
          unique_machine_count=("machine_id", "nunique"),
          unique_start_count=("start_time", "nunique"),
          unique_end_count=("end_time", "nunique"),
          unique_event_count=("event", "nunique"),
      )
      .reset_index()
)

candidate_summary["multiple_machines"] = (
    candidate_summary["unique_machine_count"] > 1
)

candidate_summary["multiple_usage_windows"] = (
    (candidate_summary["unique_start_count"] > 1)
    | (candidate_summary["unique_end_count"] > 1)
)


print("\nCandidate groups:", len(candidate_summary))

print(
    "Groups with multiple machines:",
    int(candidate_summary["multiple_machines"].sum())
)

print(
    "Groups with multiple usage windows:",
    int(candidate_summary["multiple_usage_windows"].sum())
)


print("\nLargest candidate groups:")
display(
    candidate_summary
    .sort_values("raw_row_count", ascending=False)
    .head(20)
)
```

```python
# =========================================================
# Cell 8: Compare request memory with assigned memory
# =========================================================
#
# This cell checks whether:
# - resource_request["memory"]
# - assigned_memory
#
# represent the same quantity or different memory concepts.
#
# No unit conversion or filtering is performed.
# =========================================================

memory_comparison = df[
    [
        "raw_req_memory",
        "assigned_memory",
        "page_cache_memory",
        "raw_peak_memory",
    ]
].dropna(
    subset=["raw_req_memory", "assigned_memory"]
).copy()

memory_comparison["request_minus_assigned"] = (
    memory_comparison["raw_req_memory"]
    - memory_comparison["assigned_memory"]
)

memory_comparison["request_assigned_ratio"] = np.where(
    memory_comparison["assigned_memory"] > 0,
    memory_comparison["raw_req_memory"]
    / memory_comparison["assigned_memory"],
    np.nan,
)

print("Memory comparison summary:")
display(
    memory_comparison[
        [
            "raw_req_memory",
            "assigned_memory",
            "page_cache_memory",
            "raw_peak_memory",
            "request_minus_assigned",
            "request_assigned_ratio",
        ]
    ].describe(
        percentiles=[
            0.01,
            0.25,
            0.50,
            0.75,
            0.95,
            0.99,
        ]
    ).T
)


# Check how closely request memory matches assigned memory.
print("\nRequest memory equals assigned memory:")
print(
    int(
        np.isclose(
            memory_comparison["raw_req_memory"],
            memory_comparison["assigned_memory"],
            rtol=1e-9,
            atol=1e-12,
        ).sum()
    )
)

print(
    "\nRequest memory differs from assigned memory:"
)
print(
    int(
        (
            ~np.isclose(
                memory_comparison["raw_req_memory"],
                memory_comparison["assigned_memory"],
                rtol=1e-9,
                atol=1e-12,
            )
        ).sum()
    )
)

print("\nSample memory comparisons:")
display(memory_comparison.head(10))
```
