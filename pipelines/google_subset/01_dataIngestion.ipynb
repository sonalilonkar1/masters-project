## Extract resource request and usage values

The raw CSV stores resource values as dictionary-like strings.
This step safely parses those fields and creates numeric columns for
requested, average, and peak CPU and memory.

Missing values are preserved, and no rows or events are filtered.
Units are not converted because they are handled in a later validation step.

```python
# =========================================================
# step 2: Parse nested resource fields
# =========================================================

import ast


# These columns contain dictionary-like strings.
NESTED_COLUMNS = [
    "resource_request",
    "average_usage",
    "maximum_usage",
    "random_sample_usage",
]

missing_columns = [
    column for column in NESTED_COLUMNS
    if column not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required nested columns: {missing_columns}"
    )


# Convert a dictionary-like value into a Python dictionary.
# Invalid or missing values become None.
def parse_nested(value):
    if value is None:
        return None

    if isinstance(value, float) and pd.isna(value):
        return None

    if isinstance(value, dict):
        return value

    if isinstance(value, str):
        try:
            parsed = ast.literal_eval(value)

            if isinstance(parsed, dict):
                return parsed

        except (ValueError, SyntaxError):
            pass

    return None


# Read one field from a parsed dictionary.
def get_nested_value(value, key):
    if isinstance(value, dict):
        return value.get(key)

    return None


# Parse nested columns and record missing values.
parsed_nested = {}
audit_rows = []

for column in NESTED_COLUMNS:
    parsed = df[column].map(parse_nested)
    parsed_nested[column] = parsed

    cpu_values = parsed.map(
        lambda value: get_nested_value(value, "cpus")
    )

    memory_values = parsed.map(
        lambda value: get_nested_value(value, "memory")
    )

    audit_rows.append(
        {
            "source_column": column,
            "missing_or_invalid_dict": int(parsed.isna().sum()),
            "missing_cpu_value": int(cpu_values.isna().sum()),
            "missing_memory_value": int(memory_values.isna().sum()),
        }
    )


# Store the parsing audit for review.
nested_audit = pd.DataFrame(audit_rows)

print("Nested resource-field audit:")
display(nested_audit)


# Extract numeric raw resource values.
# The raw_ prefix indicates that units have not been converted.
df["raw_req_cpu"] = pd.to_numeric(
    parsed_nested["resource_request"].map(
        lambda value: get_nested_value(value, "cpus")
    ),
    errors="coerce",
)

df["raw_req_memory"] = pd.to_numeric(
    parsed_nested["resource_request"].map(
        lambda value: get_nested_value(value, "memory")
    ),
    errors="coerce",
)

df["raw_avg_cpu"] = pd.to_numeric(
    parsed_nested["average_usage"].map(
        lambda value: get_nested_value(value, "cpus")
    ),
    errors="coerce",
)

df["raw_avg_memory"] = pd.to_numeric(
    parsed_nested["average_usage"].map(
        lambda value: get_nested_value(value, "memory")
    ),
    errors="coerce",
)

df["raw_peak_cpu"] = pd.to_numeric(
    parsed_nested["maximum_usage"].map(
        lambda value: get_nested_value(value, "cpus")
    ),
    errors="coerce",
)

df["raw_peak_memory"] = pd.to_numeric(
    parsed_nested["maximum_usage"].map(
        lambda value: get_nested_value(value, "memory")
    ),
    errors="coerce",
)


# Check the new numeric columns.
RAW_RESOURCE_COLUMNS = [
    "raw_req_cpu",
    "raw_req_memory",
    "raw_avg_cpu",
    "raw_avg_memory",
    "raw_peak_cpu",
    "raw_peak_memory",
]

print("\nExtracted raw resource values:")
display(df[RAW_RESOURCE_COLUMNS].head())

print("\nMissing percentages after extraction:")
display(
    df[RAW_RESOURCE_COLUMNS]
      .isna()
      .mean()
      .mul(100)
      .sort_values(ascending=False)
)
```

```python
# =========================================================
# Step 3: Basic resource and timestamp validation
# =========================================================
#
# This cell checks that extracted values are valid.
# It does not filter rows, convert units, or create executions.
# =========================================================

RAW_RESOURCE_COLUMNS = [
    "raw_req_cpu",
    "raw_req_memory",
    "raw_avg_cpu",
    "raw_avg_memory",
    "raw_peak_cpu",
    "raw_peak_memory",
]


# Check for negative resource values.
negative_counts = (
    df[RAW_RESOURCE_COLUMNS]
      .lt(0)
      .sum()
)

if negative_counts.any():
    raise ValueError(
        f"Negative resource values found:\n{negative_counts}"
    )


# Check that peak usage is not below average usage.
cpu_check = df[
    ["raw_avg_cpu", "raw_peak_cpu"]
].dropna()

memory_check = df[
    ["raw_avg_memory", "raw_peak_memory"]
].dropna()

if (
    cpu_check["raw_peak_cpu"]
    < cpu_check["raw_avg_cpu"]
).any():
    raise ValueError(
        "Some rows have peak CPU below average CPU."
    )

if (
    memory_check["raw_peak_memory"]
    < memory_check["raw_avg_memory"]
).any():
    raise ValueError(
        "Some rows have peak memory below average memory."
    )


# Check timestamp ordering without storing a duration yet.
raw_usage_window_delta = (
    pd.to_numeric(df["end_time"], errors="coerce")
    - pd.to_numeric(df["start_time"], errors="coerce")
)

if (raw_usage_window_delta < 0).any():
    raise ValueError(
        "Some rows have negative usage-window durations."
    )

if (raw_usage_window_delta == 0).any():
    raise ValueError(
        "Some rows have zero usage-window durations."
    )


print("Basic resource and timestamp validation passed.")
```

```python
# =========================================================
# Cell 4: Create canonical timestamp columns
# =========================================================
#
# Google trace timestamps are stored in microseconds.
# This cell creates explicit event-time and usage-window fields.
#
# The original raw timestamp columns are preserved.
# The usage-window duration is not execution runtime.
# =========================================================

INT64_MAX = 2**63 - 1


# Convert raw timestamp columns to numeric values.
event_time_us = pd.to_numeric(
    df["time"],
    errors="coerce"
)

usage_start_time_us = pd.to_numeric(
    df["start_time"],
    errors="coerce"
)

usage_end_time_us = pd.to_numeric(
    df["end_time"],
    errors="coerce"
)


# Treat the Int64 maximum value as an invalid event timestamp.
event_time_us = (
    event_time_us
    .mask(event_time_us == INT64_MAX)
    .astype("Int64")
)

usage_start_time_us = usage_start_time_us.astype("Int64")
usage_end_time_us = usage_end_time_us.astype("Int64")


# Create explicit canonical timestamp columns.
df["event_time_us"] = event_time_us
df["usage_start_time_us"] = usage_start_time_us
df["usage_end_time_us"] = usage_end_time_us


# Calculate the resource usage-window duration in seconds.
# This is not the final reconstructed execution runtime.
df["usage_window_duration_sec"] = (
    df["usage_end_time_us"]
    - df["usage_start_time_us"]
) / 1_000_000


# Validate usage-window timestamps.
if df["usage_start_time_us"].isna().any():
    raise ValueError(
        "Missing usage start timestamps found."
    )

if df["usage_end_time_us"].isna().any():
    raise ValueError(
        "Missing usage end timestamps found."
    )

if (
    df["usage_end_time_us"]
    < df["usage_start_time_us"]
).any():
    raise ValueError(
        "Negative usage-window durations found."
    )


print("Canonical timestamp columns created.")
print(
    "Invalid event timestamps converted to missing:",
    int(df["event_time_us"].isna().sum())
)

display(
    df[
        [
            "time",
            "start_time",
            "end_time",
            "event_time_us",
            "usage_start_time_us",
            "usage_end_time_us",
            "usage_window_duration_sec",
        ]
    ].head()
)
```

```python
# =========================================================
# Cell 5: Create canonical time fields
# =========================================================
# - start_time and end_time are in microseconds.
# - The trace spans approximately one month.
# - Duration is converted to seconds.
#
# The original raw timestamp columns are preserved.
# Rows are not filtered in this cell.
# =========================================================

TIMESTAMP_UNIT = "microseconds"
MICROSECONDS_PER_SECOND = 1_000_000

# Preserve raw timestamp values and ensure numeric representation.
df["start_time_us"] = pd.to_numeric(
    df["start_time"],
    errors="coerce"
)

df["end_time_us"] = pd.to_numeric(
    df["end_time"],
    errors="coerce"
)

# Calculate duration in seconds using the verified unit.
df["duration_sec"] = (
    df["end_time_us"] - df["start_time_us"]
) / MICROSECONDS_PER_SECOND

# Validate timestamp conversion.
invalid_timestamp_count = int(
    df[["start_time_us", "end_time_us"]]
    .isna()
    .any(axis=1)
    .sum()
)

negative_duration_count = int(
    (df["duration_sec"] < 0).sum()
)

zero_duration_count = int(
    (df["duration_sec"] == 0).sum()
)

print("Timestamp unit:", TIMESTAMP_UNIT)
print("Invalid timestamp rows:", invalid_timestamp_count)
print("Negative-duration rows:", negative_duration_count)
print("Zero-duration rows:", zero_duration_count)

print("\nDuration in seconds:")
display(
    df["duration_sec"].describe(
        percentiles=[0.01, 0.50, 0.95, 0.99]
    )
)

print("\nCanonical time fields:")
display(
    df[
        [
            "start_time",
            "end_time",
            "start_time_us",
            "end_time_us",
            "duration_sec",
        ]
    ].head()
)
```

```python
# =========================================================
# Cell 6: Audit CPU and memory value scales
# =========================================================
#
# No unit conversion is performed in this cell.
# The values remain in their original raw scale.
# =========================================================

RESOURCE_SCALE_COLUMNS = [
    "raw_req_cpu",
    "raw_avg_cpu",
    "raw_peak_cpu",
    "raw_req_memory",
    "raw_avg_memory",
    "raw_peak_memory",
    "assigned_memory",
    "page_cache_memory",
]

# Display resource distributions.
print("Resource value summary:")

display(
    df[RESOURCE_SCALE_COLUMNS]
    .describe(
        percentiles=[
            0.01,
            0.25,
            0.50,
            0.75,
            0.95,
            0.99,
            0.999,
        ]
    )
    .T
)


# Count values greater than one.
# This is only a diagnostic; values above one are not
# automatically incorrect.
print("\nValues greater than 1:")

display(
    df[RESOURCE_SCALE_COLUMNS]
    .gt(1)
    .sum()
    .sort_values(ascending=False)
)


# Verify that peak CPU is not lower than average CPU.
cpu_check = df[
    ["raw_avg_cpu", "raw_peak_cpu"]
].dropna()

print("\nCPU relationship:")
print("Rows compared:", len(cpu_check))
print(
    "Peak CPU >= average CPU:",
    int(
        (
            cpu_check["raw_peak_cpu"]
            >= cpu_check["raw_avg_cpu"]
        ).sum()
    )
)
print(
    "Peak CPU < average CPU:",
    int(
        (
            cpu_check["raw_peak_cpu"]
            < cpu_check["raw_avg_cpu"]
        ).sum()
    )
)


# Verify that peak memory is not lower than average memory.
memory_check = df[
    ["raw_avg_memory", "raw_peak_memory"]
].dropna()

print("\nMemory relationship:")
print("Rows compared:", len(memory_check))
print(
    "Peak memory >= average memory:",
    int(
        (
            memory_check["raw_peak_memory"]
            >= memory_check["raw_avg_memory"]
        ).sum()
    )
)
print(
    "Peak memory < average memory:",
    int(
        (
            memory_check["raw_peak_memory"]
            < memory_check["raw_avg_memory"]
        ).sum()
    )
)


# Compare request and peak values.
# These are preliminary diagnostics, not final violations.
cpu_request_check = df[
    ["raw_req_cpu", "raw_peak_cpu"]
].dropna()

memory_request_check = df[
    ["raw_req_memory", "raw_peak_memory"]
].dropna()

print("\nRequest versus peak:")
print(
    "CPU rows where peak > request:",
    int(
        (
            cpu_request_check["raw_peak_cpu"]
            > cpu_request_check["raw_req_cpu"]
        ).sum()
    )
)

print(
    "Memory rows where peak > request:",
    int(
        (
            memory_request_check["raw_peak_memory"]
            > memory_request_check["raw_req_memory"]
        ).sum()
    )
)
```

```python
# =========================================================
# Cell 7: Compare request memory with assigned memory
# =========================================================
#
# This cell checks whether:
# - resource_request["memory"]
# - assigned_memory
#
# represent the same quantity or different memory concepts.
#
# No unit conversion or filtering is performed.
# =========================================================

memory_comparison = df[
    [
        "raw_req_memory",
        "assigned_memory",
        "page_cache_memory",
        "raw_peak_memory",
    ]
].dropna(
    subset=["raw_req_memory", "assigned_memory"]
).copy()

memory_comparison["request_minus_assigned"] = (
    memory_comparison["raw_req_memory"]
    - memory_comparison["assigned_memory"]
)

memory_comparison["request_assigned_ratio"] = np.where(
    memory_comparison["assigned_memory"] > 0,
    memory_comparison["raw_req_memory"]
    / memory_comparison["assigned_memory"],
    np.nan,
)

print("Memory comparison summary:")
display(
    memory_comparison[
        [
            "raw_req_memory",
            "assigned_memory",
            "page_cache_memory",
            "raw_peak_memory",
            "request_minus_assigned",
            "request_assigned_ratio",
        ]
    ].describe(
        percentiles=[
            0.01,
            0.25,
            0.50,
            0.75,
            0.95,
            0.99,
        ]
    ).T
)


# Check how closely request memory matches assigned memory.
print("\nRequest memory equals assigned memory:")
print(
    int(
        np.isclose(
            memory_comparison["raw_req_memory"],
            memory_comparison["assigned_memory"],
            rtol=1e-9,
            atol=1e-12,
        ).sum()
    )
)

print(
    "\nRequest memory differs from assigned memory:"
)
print(
    int(
        (
            ~np.isclose(
                memory_comparison["raw_req_memory"],
                memory_comparison["assigned_memory"],
                rtol=1e-9,
                atol=1e-12,
            )
        ).sum()
    )
)

print("\nSample memory comparisons:")
display(memory_comparison.head(10))
```

## Candidate Execution Identity Check

This analysis tested whether `collection_id + instance_index` uniquely identifies one workload execution.

The dataset contains:

- 405,894 raw rows;
- 242,946 candidate execution groups;
- median of 1 raw row per candidate group;
- maximum of 7,704 raw rows for one candidate group.

Some candidate groups contain many repeated rows and multiple event types across a large time range. For example, one candidate group contained 7,704 rows with repeated `ENABLE` and `FAIL` events.

### Finding

`collection_id + instance_index` cannot yet be treated as the final `execution_id`.

It may represent a recurring task or instance key rather than one individual execution. We must inspect repeated groups and determine how separate execution episodes should be identified before aggregating CPU, memory, duration, or targets.

This decision affects:

- execution reconstruction;
- peak CPU and memory aggregation;
- recurrence grouping;
- historical profiles;
- target construction;
- baseline evaluation.

```python
# =========================================================
# Cell 11: Inspect repeated candidate execution groups
# =========================================================
#
# This cell investigates candidate IDs with many raw rows.
# It helps determine whether the candidate key represents
# one execution or multiple repeated episodes.
#
# No rows are filtered or modified.
# =========================================================

# Select the candidate groups with the most raw rows.
largest_candidate_ids = (
    candidate_execution_summary["raw_row_count"]
    .sort_values(ascending=False)
    .head(5)
    .index
)

for candidate_id in largest_candidate_ids:
    print("\n" + "=" * 80)
    print("Candidate execution:", candidate_id)

    candidate_rows = (
        df[df["_candidate_execution_id"] == candidate_id]
        .sort_values(
            ["start_time_us", "end_time_us", "time"],
            kind="mergesort"
        )
    )

    print("Raw rows:", len(candidate_rows))
    print("Event types:", candidate_rows["event"].unique())
    print("Machines:", candidate_rows["machine_id"].nunique())
    print("Distinct starts:", candidate_rows["start_time_us"].nunique())
    print("Distinct ends:", candidate_rows["end_time_us"].nunique())

    display(
        candidate_rows[
            [
                "_candidate_execution_id",
                "collection_id",
                "instance_index",
                "machine_id",
                "event",
                "failed",
                "time",
                "start_time_us",
                "end_time_us",
                "duration_sec",
                "raw_req_cpu",
                "raw_req_memory",
                "raw_peak_cpu",
                "raw_peak_memory",
            ]
        ].head(30)
    )
```

```python
# =========================================================
# Cell 12: Inspect candidate event/time episodes
# =========================================================
#
# This cell checks whether adding the event and time interval
# separates the broad candidate groups into smaller episodes.
#
# These are diagnostic episode groups only.
# They are not yet the final execution IDs.
# =========================================================

episode_summary = (
    df.groupby(
        [
            "_candidate_execution_id",
            "start_time_us",
            "end_time_us",
            "event",
        ],
        dropna=False,
    )
    .agg(
        raw_row_count=("event", "size"),
        machine_count=("machine_id", "nunique"),
        distinct_time_count=("time", "nunique"),
        failed_row_count=("failed", "sum"),
        peak_cpu_min=("raw_peak_cpu", "min"),
        peak_cpu_max=("raw_peak_cpu", "max"),
        peak_memory_min=("raw_peak_memory", "min"),
        peak_memory_max=("raw_peak_memory", "max"),
    )
    .reset_index()
)

print("Episode groups:", len(episode_summary))

print("\nEpisode-group summary:")
display(
    episode_summary[
        [
            "raw_row_count",
            "machine_count",
            "distinct_time_count",
            "failed_row_count",
        ]
    ].describe()
)


print("\nLargest episode groups:")
display(
    episode_summary
    .sort_values("raw_row_count", ascending=False)
    .head(20)
)


print("\nCandidate groups versus episode groups:")
print(
    "Candidate groups:",
    df["_candidate_execution_id"].nunique()
)
print(
    "Episode groups:",
    len(episode_summary)
)
```

## Raw Observation Key

The combination of `collection_id`, `instance_index`, `machine_id`,
`start_time`, and `end_time` uniquely identifies all 405,894 raw rows.

Therefore, the raw dataset contains unique observation records at this
machine-and-time-window grain.

This key is not yet the modeling execution key. Multiple observation records
may still belong to the same reconstructed workload execution.

```python
# Examine the structure of each collection_id + instance_index group.
# This is diagnostic only; it does not create the final execution_id.

candidate_summary = (
    df.groupby(["collection_id", "instance_index"], dropna=False)
      .agg(
          raw_row_count=("machine_id", "size"),
          unique_machine_count=("machine_id", "nunique"),
          unique_start_count=("start_time", "nunique"),
          unique_end_count=("end_time", "nunique"),
          unique_event_count=("event", "nunique"),
          first_start_time=("start_time", "min"),
          last_end_time=("end_time", "max"),
      )
      .reset_index()
)

# Convert the overall time span to seconds for easier interpretation.
candidate_summary["time_span_sec"] = (
    candidate_summary["last_end_time"]
    - candidate_summary["first_start_time"]
) / 1_000_000

# Identify groups where all machines share the same observation window.
candidate_summary["same_observation_window"] = (
    (candidate_summary["unique_start_count"] == 1)
    & (candidate_summary["unique_end_count"] == 1)
)

print("Candidate groups:", len(candidate_summary))

print("\nCandidate groups with multiple machines:")
print(
    (candidate_summary["unique_machine_count"] > 1)
    .value_counts()
)

print("\nCandidate groups with multiple time windows:")
print(
    (
        (candidate_summary["unique_start_count"] > 1)
        | (candidate_summary["unique_end_count"] > 1)
    ).value_counts()
)

print("\nLargest candidate groups:")
display(
    candidate_summary
    .sort_values(
        ["raw_row_count", "unique_machine_count"],
        ascending=False
    )
    .head(20)
)
```

```python
# Check whether multi-machine groups contain identical resource and event values.
# The grouping excludes machine_id so that all machines in the same
# collection, instance, and observation window are compared together.

window_key = [
    "collection_id",
    "instance_index",
    "start_time",
    "end_time",
]

window_summary = (
    df.groupby(window_key, dropna=False)
      .agg(
          raw_row_count=("machine_id", "size"),
          unique_machine_count=("machine_id", "nunique"),
          unique_event_count=("event", "nunique"),
          unique_req_cpu=("raw_req_cpu", "nunique"),
          unique_req_memory=("raw_req_memory", "nunique"),
          unique_peak_cpu=("raw_peak_cpu", "nunique"),
          unique_peak_memory=("raw_peak_memory", "nunique"),
      )
      .reset_index()
)

multi_machine_windows = window_summary[
    window_summary["unique_machine_count"] > 1
].copy()

print("Multi-machine observation windows:", len(multi_machine_windows))

print("\nResource/event consistency among multi-machine windows:")
display(
    multi_machine_windows[
        [
            "unique_event_count",
            "unique_req_cpu",
            "unique_req_memory",
            "unique_peak_cpu",
            "unique_peak_memory",
        ]
    ]
    .value_counts()
    .head(20)
)

print("\nLargest multi-machine windows:")
display(
    multi_machine_windows
    .sort_values(
        ["unique_machine_count", "raw_row_count"],
        ascending=False
    )
    .head(20)
)
```

```python
# Inspect the largest multi-machine observation window row by row.
# This helps determine whether resource values vary across machines.

largest_window = (
    multi_machine_windows
    .sort_values(
        ["unique_machine_count", "raw_row_count"],
        ascending=False
    )
    .iloc[0]
)

window_mask = (
    (df["collection_id"] == largest_window["collection_id"])
    & (df["instance_index"] == largest_window["instance_index"])
    & (df["start_time"] == largest_window["start_time"])
    & (df["end_time"] == largest_window["end_time"])
)

inspection_cols = [
    "collection_id",
    "instance_index",
    "machine_id",
    "event",
    "failed",
    "start_time",
    "end_time",
    "raw_req_cpu",
    "raw_req_memory",
    "raw_avg_cpu",
    "raw_avg_memory",
    "raw_peak_cpu",
    "raw_peak_memory",
]

print("Rows in selected window:", int(window_mask.sum()))

display(
    df.loc[window_mask, inspection_cols]
      .sort_values(
          ["raw_req_cpu", "raw_req_memory", "machine_id"]
      )
      .head(30)
)

print("\nDistinct values in selected window:")

for col in [
    "event",
    "failed",
    "raw_req_cpu",
    "raw_req_memory",
    "raw_avg_cpu",
    "raw_avg_memory",
    "raw_peak_cpu",
    "raw_peak_memory",
]:
    print(f"\n{col}:")
    print(df.loc[window_mask, col].value_counts(dropna=False).head(10))
```

```python
# Inspect time and source metadata for the selected multi-machine window.
# This helps determine whether machine rows correspond to distinct
# observation timestamps or repeated source records.

selected = df.loc[window_mask].copy()

window_start = selected["start_time"].iloc[0]
window_end = selected["end_time"].iloc[0]

selected["time_inside_observation_window"] = (
    (selected["time"] >= window_start)
    & (selected["time"] <= window_end)
)

print("Distinct machine IDs:", selected["machine_id"].nunique())
print("Distinct raw time values:", selected["time"].nunique())
print("Minimum raw time:", selected["time"].min())
print("Maximum raw time:", selected["time"].max())
print(
    "Rows with raw time inside observation window:",
    int(selected["time_inside_observation_window"].sum())
)

print("\nRequest/resource combinations:")
display(
    selected.groupby(
        [
            "raw_req_cpu",
            "raw_req_memory",
            "raw_peak_cpu",
            "raw_peak_memory",
            "event",
            "failed",
        ],
        dropna=False
    )
    .size()
    .reset_index(name="row_count")
    .sort_values("row_count", ascending=False)
)

print("\nMetadata consistency:")
metadata_cols = [
    "collection_type",
    "priority",
    "alloc_collection_id",
    "user",
    "collection_name",
    "collection_logical_name",
    "scheduler",
    "vertical_scaling",
]

display(
    selected[metadata_cols]
    .nunique(dropna=False)
    .sort_values(ascending=False)
)

print("\nSample rows ordered by raw time:")
display(
    selected.sort_values("time")[
        [
            "time",
            "machine_id",
            "event",
            "raw_req_cpu",
            "raw_req_memory",
            "raw_peak_cpu",
            "raw_peak_memory",
        ]
    ].head(30)
)
```

## Time and Event-Usage Alignment Issue

For a selected multi-machine group, all rows have `time = 0`, while
`start_time` and `end_time` define a later usage window.

The rows also contain thousands of machine IDs with identical event and
peak-usage values. This indicates that the CSV may combine event-level and
usage-level fields at different grains.

The raw `time` field must not be used for execution ordering until its
relationship with `start_time`, `end_time`, and event type is verified.

```python
# Compare the raw event time with the usage observation window.
# This checks whether time is aligned with start_time/end_time
# for each event type.

time_alignment = (
    df.groupby("event", dropna=False)
      .agg(
          row_count=("event", "size"),
          zero_time_rows=("time", lambda x: (x == 0).sum()),
          distinct_time_values=("time", "nunique"),
          distinct_start_values=("start_time", "nunique"),
          distinct_end_values=("end_time", "nunique"),
      )
      .reset_index()
)

time_alignment["zero_time_percentage"] = (
    time_alignment["zero_time_rows"]
    / time_alignment["row_count"]
    * 100
)

display(
    time_alignment.sort_values(
        "row_count",
        ascending=False
    )
)

# Check whether raw time falls inside the usage window.
df["time_inside_usage_window"] = (
    (df["time"] >= df["start_time"])
    & (df["time"] <= df["end_time"])
)

print(
    "Rows with time inside usage window:",
    int(df["time_inside_usage_window"].sum()),
    "of",
    len(df)
)

print(
    "Percentage:",
    f"{df['time_inside_usage_window'].mean() * 100:.2f}%"
)
```
